### Número de placa, patrulla asignada, sector de vigilancia, años de servicio, rango y curso de especialización.

```python
# =========================
# MENÚ DE POLICÍAS
# =========================

diccionario_policia = {}

# =========================
# FUNCIÓN AGREGAR
# =========================
def agregar_policia():

    clave = input("Ingresa el número de placa: ")

    patrulla = input("Ingresa la patrulla asignada: ")
    sector = input("Ingresa el sector de vigilancia: ")
    anios = input("Ingresa los años de servicio: ")
    rango = input("Ingresa el rango: ")
    curso = input("Ingresa el curso de especialización: ")

    diccionario_policia[clave] = {
        "patrulla": patrulla,
        "sector": sector,
        "años": anios,
        "rango": rango,
        "curso": curso
    }

    print("Policía agregado correctamente")


# =========================
# FUNCIÓN LISTAR
# =========================
def Listar_policia():

    if diccionario_policia:

        print("\n==============================================================================================")
        print("{:<12} {:<20} {:<20} {:<15} {:<15} {:<25}".format(
            "PLACA",
            "PATRULLA",
            "SECTOR",
            "AÑOS",
            "RANGO",
            "CURSO"
        ))
        print("==============================================================================================")

        for key, value in diccionario_policia.items():

            print("{:<12} {:<20} {:<20} {:<15} {:<15} {:<25}".format(
                key,
                value["patrulla"],
                value["sector"],
                value["años"],
                value["rango"],
                value["curso"]
            ))

    else:
        print("No hay policías registrados")


# =========================
# FUNCIÓN BUSCAR
# =========================
def buscar_policia():

    clave = input("Ingresa el número de placa a buscar: ")

    if clave in diccionario_policia:

        print("\n===== POLICÍA ENCONTRADO =====")

        for key, value in diccionario_policia[clave].items():
            print(key, ":", value)

    else:
        print("Policía no encontrado")


# =========================
# FUNCIÓN BORRAR
# =========================
def borrar_policia():

    clave = input("Ingresa el número de placa a borrar: ")

    if clave in diccionario_policia:

        del diccionario_policia[clave]

        print("Policía eliminado correctamente")

    else:
        print("Policía no encontrado")


# =========================
# FUNCIÓN ACTUALIZAR
# =========================
def actualizar_policia():

    clave = input("Ingresa el número de placa a actualizar: ")

    if clave in diccionario_policia:

        patrulla = input("Nueva patrulla asignada: ")
        sector = input("Nuevo sector de vigilancia: ")
        anios = input("Nuevos años de servicio: ")
        rango = input("Nuevo rango: ")
        curso = input("Nuevo curso de especialización: ")

        diccionario_policia[clave] = {
            "patrulla": patrulla,
            "sector": sector,
            "años": anios,
            "rango": rango,
            "curso": curso
        }

        print("Policía actualizado correctamente")

    else:
        print("Policía no encontrado")


# =========================
# MENÚ PRINCIPAL
# =========================

while True:

    print("\n===== MENÚ =====")
    print("1. Agregar policía en un diccionario")
    print("2. Listar diccionarios de policía")
    print("3. Buscar policía en diccionario")
    print("4. Borrar policía en diccionario")
    print("5. Actualizar policía en diccionario")
    print("6. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        agregar_policia()

    elif opcion == "2":
        Listar_policia()

    elif opcion == "3":
        buscar_policia()

    elif opcion == "4":
        borrar_policia()

    elif opcion == "5":
        actualizar_policia()

    elif opcion == "6":
        print("Programa finalizado")
        break

    else:
        print("Opción no válida")


print("Programa realizado por Alejandra Rivera Valenzuela NL 45")
```
