### Número de placa, patrulla asignada, sector de vigilancia, años de servicio, rango y curso de especialización.

```python
# =====================================================
# PROYECTO: SISTEMA DE POLICÍAS
# =====================================================

# Diccionario vacío
diccionario_policia = []


# =====================================================
# FUNCIÓN LISTAR POLICÍAS
# =====================================================

def listar_policia():

    print("\n================ LISTA DE POLICÍAS ================\n")

    print("{:<12} {:<20} {:<20} {:<10} {:<15} {:<25}".format(
        "PLACA",
        "PATRULLA",
        "SECTOR",
        "AÑOS",
        "RANGO",
        "CURSO"
    ))

    print("-" * 110)

    for policia in diccionario_policia:

        print("{:<12} {:<20} {:<20} {:<10} {:<15} {:<25}".format(
            policia["Placa"],
            policia["Patrulla"],
            policia["Sector"],
            policia["Años"],
            policia["Rango"],
            policia["Curso"]
        ))


# =====================================================
# FUNCIÓN AGREGAR POLICÍA
# =====================================================

def agregar_policia():

    print("\n============= AGREGAR POLICÍA =============\n")

    placa = input("Número de placa: ")
    patrulla = input("Patrulla asignada: ")
    sector = input("Sector de vigilancia: ")
    anios = input("Años de servicio: ")
    rango = input("Rango: ")
    curso = input("Curso de especialización: ")

    policia = {
        "Placa": placa,
        "Patrulla": patrulla,
        "Sector": sector,
        "Años": anios,
        "Rango": rango,
        "Curso": curso
    }

    diccionario_policia.append(policia)

    print("\nPOLICÍA AGREGADO CORRECTAMENTE")

    listar_policia()


# =====================================================
# FUNCIÓN BUSCAR POLICÍA
# =====================================================

def buscar_policia():

    print("\n============= BUSCAR POLICÍA =============\n")

    placa_buscar = input("Ingrese número de placa: ")

    for policia in diccionario_policia:

        if policia["Placa"] == placa_buscar:

            print("\nPOLICÍA ENCONTRADO\n")

            for key, value in policia.items():
                print(key, ":", value)

            return

    print("\nNo se encontró el policía")


# =====================================================
# FUNCIÓN BORRAR POLICÍA
# =====================================================

def borrar_policia():

    print("\n============= BORRAR POLICÍA =============\n")

    placa_borrar = input("Ingrese número de placa a borrar: ")

    for policia in diccionario_policia:

        if policia["Placa"] == placa_borrar:

            diccionario_policia.remove(policia)

            print("\nPOLICÍA BORRADO CORRECTAMENTE")

            listar_policia()

            return

    print("\nNo se encontró el policía")


# =====================================================
# FUNCIÓN ACTUALIZAR POLICÍA
# =====================================================

def actualizar_policia():

    print("\n============= ACTUALIZAR POLICÍA =============\n")

    placa_actualizar = input("Ingrese número de placa: ")

    for policia in diccionario_policia:

        if policia["Placa"] == placa_actualizar:

            policia["Patrulla"] = input("Nueva patrulla: ")
            policia["Sector"] = input("Nuevo sector: ")
            policia["Años"] = input("Nuevos años de servicio: ")
            policia["Rango"] = input("Nuevo rango: ")
            policia["Curso"] = input("Nuevo curso: ")

            print("\nPOLICÍA ACTUALIZADO CORRECTAMENTE")

            listar_policia()

            return

    print("\nNo se encontró el policía")


# =====================================================
# MENÚ PRINCIPAL
# =====================================================

while True:

    print("\n================ MENÚ =================")
    print("1. Agregar policía")
    print("2. Listar policías")
    print("3. Buscar policía")
    print("4. Borrar policía")
    print("5. Actualizar policía")
    print("6. Salir")

    opcion = input("\nSeleccione una opción: ")


    if opcion == "1":
        agregar_policia()


    elif opcion == "2":
        listar_policia()


    elif opcion == "3":
        buscar_policia()
        listar_policia()


    elif opcion == "4":
        borrar_policia()


    elif opcion == "5":
        actualizar_policia()


    elif opcion == "6":
        print("\nPrograma finalizado")
        break


    else:
        print("\nOpción incorrecta")
        print("Realizado por Alejandra Rivera Valenzuela NL = 45 ")
```
